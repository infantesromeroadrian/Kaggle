```python
import os, shutil, subprocess, glob
from pathlib import Path

OUTPUT_DIR = Path('/kaggle/working')

# Find adapter files
adapter_config = None
adapter_model = None
for p in glob.glob('/kaggle/input/**/adapter_config.json', recursive=True):
    adapter_config = p
for p in glob.glob('/kaggle/input/**/adapter_model.safetensors', recursive=True):
    adapter_model = p

if adapter_config and adapter_model:
    shutil.copy2(adapter_config, OUTPUT_DIR / 'adapter_config.json')
    shutil.copy2(adapter_model, OUTPUT_DIR / 'adapter_model.safetensors')
    os.chdir(str(OUTPUT_DIR))
    subprocess.run(['zip', 'submission.zip', 'adapter_config.json', 'adapter_model.safetensors'], check=True)
    print(f'submission.zip: {os.path.getsize("submission.zip")/1024/1024:.1f} MB')
    print('Done!')
else:
    print(f'ERROR: adapter_config={adapter_config}, adapter_model={adapter_model}')
    for d in sorted(Path('/kaggle/input').rglob('*')):
        if d.is_file():
            print(f'  {d} ({d.stat().st_size/1024:.1f} KB)')
```
